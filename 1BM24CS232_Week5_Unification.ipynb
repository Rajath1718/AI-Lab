```python

import re

# Check whether a term is a variable
def is_variable(term):
    return term.islower() and term.isalpha()

# Apply substitutions to a term
def substitute(term, subs):
    while term in subs:
        term = subs[term]
    return term

# Check occurs condition
def occurs(var, term, subs):
    term = substitute(term, subs)

    if var == term:
        return True

    return False

# Unification of two terms
def unify_terms(t1, t2, subs):

    t1 = substitute(t1, subs)
    t2 = substitute(t2, subs)

    # If both terms are identical
    if t1 == t2:
        return True

    # If first term is a variable
    if is_variable(t1):
        if occurs(t1, t2, subs):
            return False
        subs[t1] = t2
        return True

    # If second term is a variable
    if is_variable(t2):
        if occurs(t2, t1, subs):
            return False
        subs[t2] = t1
        return True

    # Both are different constants
    return False

# Parse predicate expressions
def parse_expression(expr):
    match = re.fullmatch(r'\s*(\w+)\s*\((.*)\)\s*', expr)

    if not match:
        raise ValueError("Invalid expression format")

    predicate = match.group(1)
    arguments = [arg.strip() for arg in match.group(2).split(',')]

    if not all(arguments):
        raise ValueError("Invalid arguments")

    return predicate, arguments

# Main unification algorithm
def unify(expr1, expr2):

    predicate1, args1 = parse_expression(expr1)
    predicate2, args2 = parse_expression(expr2)

    subs = {}

    print("\nExpression 1:", expr1)
    print("Expression 2:", expr2)

    # Step 1: Check predicate symbols
    if predicate1 != predicate2:
        print("\nUnification FAILED: Different predicates.")
        return

    # Step 2: Check number of arguments
    if len(args1) != len(args2):
        print("\nUnification FAILED: Different number of arguments.")
        return

    # Step 3: Unify corresponding arguments
    for a, b in zip(args1, args2):
        if not unify_terms(a, b, subs):
            print("\nUnification FAILED: Incompatible terms.")
            return

    # Step 4: Display substitutions
    print("\nUnification SUCCESSFUL!")
    print("Substitution Set:")

    for var, value in subs.items():
        print(var, "=", substitute(value, subs))

    # Step 5: Apply substitutions
    def apply_all(args):
        return [substitute(arg, subs) for arg in args]

    unified_args = apply_all(args1)

    print("\nUnified Expression:")
    print(predicate1 + "(" + ", ".join(unified_args) + ")")


# Main program
print("UNIFICATION IN FIRST ORDER LOGIC")

e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

unify(e1, e2)
```
