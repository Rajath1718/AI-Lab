```python

# Alpha-Beta Pruning Algorithm

# Game tree from the given PPT
tree = [
    [[10, 9], [14, 18]],
    [[5, 4], [50, 3]]
]

pruned = []

def alpha_beta(node, alpha, beta, maximizing, path="Root"):

    # If leaf node, return its value
    if isinstance(node, int):
        print(f"Visited {path}: {node}, Alpha={alpha}, Beta={beta}")
        return node, [path]

    if maximizing:
        value = float('-inf')
        best_path = []

        for i, child in enumerate(node):
            new_path = path + f" -> {i+1}"

            result, result_path = alpha_beta(
                child, alpha, beta, False, new_path
            )

            if result > value:
                value = result
                best_path = result_path

            alpha = max(alpha, value)

            # Pruning condition
            if alpha >= beta:
                for j in range(i + 1, len(node)):
                    pruned.append(
                        (path + f" -> {j+1}", node[j])
                    )
                break

        return value, best_path

    else:
        value = float('inf')
        best_path = []

        for i, child in enumerate(node):
            new_path = path + f" -> {i+1}"

            result, result_path = alpha_beta(
                child, alpha, beta, True, new_path
            )

            if result < value:
                value = result
                best_path = result_path

            beta = min(beta, value)

            # Pruning condition
            if alpha >= beta:
                for j in range(i + 1, len(node)):
                    pruned.append(
                        (path + f" -> {j+1}", node[j])
                    )
                break

        return value, best_path


# Execute Alpha-Beta Pruning
print("Alpha-Beta Pruning Search:\n")

result, best_path = alpha_beta(
    tree, float('-inf'), float('inf'), True
)

print("\nFinal MAX Root Value:", result)
print("Optimal Path:", " -> ".join(best_path))

print("\nPruned Subtrees:")
for path, subtree in pruned:
    print(path, ":", subtree)
```
