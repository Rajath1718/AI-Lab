```python

import random
import math

N = 8

# Calculate number of attacking queen pairs
def calculate_cost(board):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):

            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                cost += 1

    return cost


# Generate a random neighbouring state
def get_neighbor(board):
    new_board = board.copy()

    col = random.randint(0, N - 1)
    new_row = random.randint(0, N - 1)

    while new_row == new_board[col]:
        new_row = random.randint(0, N - 1)

    new_board[col] = new_row

    return new_board


# Simulated Annealing Algorithm
def simulated_annealing():

    current = [random.randint(0, N - 1) for _ in range(N)]

    T = 10.0
    cooling_rate = 0.99
    min_temperature = 0.001

    iteration = 0

    while T > min_temperature:

        current_cost = calculate_cost(current)

        # Check if solution is found
        if current_cost == 0:
            return current, iteration

        # Generate neighbouring state
        next_state = get_neighbor(current)
        next_cost = calculate_cost(next_state)

        # Calculate change in energy
        delta_E = current_cost - next_cost

        # Accept better solution
        if delta_E > 0:
            current = next_state

        else:
            # Accept worse solution with probability
            probability = math.exp(delta_E / T)

            if random.random() < probability:
                current = next_state

        # Decrease temperature
        T = T * cooling_rate
        iteration += 1

    return current, iteration


# Display chessboard
def display_board(board):

    for row in range(N):
        for col in range(N):

            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")

        print()


# Main Program
random.seed(42)

for attempt in range(100):

    solution, iterations = simulated_annealing()

    if calculate_cost(solution) == 0:

        print("\nSolution Found!")
        print("Number of Attempts:", attempt + 1)
        print("Iterations:", iterations)

        # Display final cost
        print("Final Cost:", calculate_cost(solution))

        # Display chessboard
        print("\nChessboard:")
        display_board(solution)

        # Display final queen positions as a list
        print("\nFinal Solution:")
        print("Final Queen Positions:",
              [row + 1 for row in solution])

        break

else:
    print("No solution found. Try again.")
```
